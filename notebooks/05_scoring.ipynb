# Preparations
---

## Environment Setup

```python
import sys
import os
from pathlib import Path
import pandas as pd

PROJECT_ROOT = os.path.abspath("..")
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

print("PYTHONPATH OK:", PROJECT_ROOT)
```

## Import Scoring Engine

```python
from src._6_scoring.scoring_engine import run_scoring_pipeline
```

# Load reconciliation output
---

```python
anomalies_summary = pd.read_parquet("../data/_3_gold/reconciliation/anomalies_summary.parquet")

print("Reconciliation summary loaded.")
display(anomalies_summary.head())
```

# Run DQI Scoring Pipeline
---

```python
print("\n## Running DQI scoring pipeline...")

scoring_output = run_scoring_pipeline(anomalies_summary)

print("\nPipeline completed.")
```

# Display Results
---

```python
entity_scores = scoring_output["entity_scores"]
dataset_score = scoring_output["dataset_score"]

print("\n### Entity Scores (first 20 rows)")
display(entity_scores.head(20))

print("\n### Dataset Score")
dataset_score
```

# Summary
---

```python
print("\n--- DQI SCORING SUMMARY ---")
print(f"Entities scored: {len(entity_scores)}")
print(f"Global DQI score: {dataset_score['global_score']} ({dataset_score['global_label']})")
print(f"Weighted global score: {dataset_score['weighted_global_score']} ({dataset_score['weighted_global_label']})")
print(f"Severity distribution: {dataset_score['severity_distribution']}")
print(f"Anomaly density: {dataset_score['anomaly_density']}")
print(f"Source reliability: {dataset_score['sources_reliability']}")
print(f"Entity-type contribution: {dataset_score['entity_type_contribution']}")

print("\nDataset scored. Proceed to Notebook 06 — Dashboard Export.")
```

# Save Outputs
---

```python
import json

output_dir = "../data/_3_gold/scoring/"
os.makedirs(output_dir, exist_ok=True)

# Save entity scores
entity_scores.to_parquet(f"{output_dir}/entity_scores.parquet", index=False)

# Save dataset score (JSON)
with open(f"{output_dir}/dataset_score.json", "w") as f:
    json.dump(dataset_score, f, indent=4)

print("Saved scoring outputs.")
```
